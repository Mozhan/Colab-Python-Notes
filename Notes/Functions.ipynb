## Understanding Function Arguments in Python

To understand how to pass information to functions, we use different types of arguments:

1. **Positional Arguments**: Arguments that must be passed in the correct positional order. The first argument in the function call maps to the first parameter, the second to the second, and so on.
2. **Keyword Arguments**: Arguments passed by explicitly specifying the parameter name and its value (e.g., `name="Alice"`). Order does not matter when using keyword arguments.
3. **Default Arguments**: Parameters in a function definition that have a fallback value. If no argument is provided for that parameter during the function call, the default value is used.

### Example 1: Positional Arguments
In this example, the values are assigned to the parameters strictly based on their order.

```python
def describe_pet(animal_type, pet_name):
    print(f"I have a {animal_type} named {pet_name}.")

# Positional arguments: "hamster" goes to animal_type, "Harry" goes to pet_name
describe_pet("hamster", "Harry")
```

### Example 2: Keyword Arguments
Here, we explicitly use parameter names, which allows us to pass them in any order.

```python
def describe_pet(animal_type, pet_name):
    print(f"I have a {animal_type} named {pet_name}.")

# Keyword arguments: order does not matter
describe_pet(pet_name="Lassie", animal_type="dog")
```

### Example 3: Default Arguments
If a parameter is defined with a default value, passing an argument for it during the call is optional.

```python
def describe_pet(pet_name, animal_type="dog"):
    print(f"I have a {animal_type} named {pet_name}.")

# Using the default argument for animal_type
describe_pet(pet_name="Rex")

# Overriding the default argument
describe_pet(pet_name="Whiskers", animal_type="cat")
```

## Understanding Docstrings in Python

A **docstring** (documentation string) is a special string literal placed as the very first statement in a function, class, module, or method definition. It is used to document what the code does, its arguments, and its return values.

### 1. How to Create a Docstring
You can create a docstring using triple quotes (`"""` or `'''`). This allows the docstring to span multiple lines.

```python
def calculate_miles(kilometers):
    """
    Convert kilometers to miles.

    Parameters:
    kilometers (float): The distance in kilometers.

    Returns:
    float: The converted distance in miles.
    """
    return kilometers * 0.621371
```

### 2. How to Access a Docstring
You can access a function's docstring programmatically using the `__doc__` attribute, or interactively using Python's built-in `help()` function.

```python
# Accessing via the __doc__ attribute
print("Using __doc__:\n")
print(calculate_miles.__doc__)

print("-" * 40)

# Accessing via the help() function
print("Using help():\n")
help(calculate_miles)
```

### 3. How to Modify a Docstring
Because docstrings are stored in the runtime object's `__doc__` attribute, you can modify them dynamically by reassigning that attribute.

```python
# Modifying the docstring dynamically
calculate_miles.__doc__ = "New docstring: Converts km to miles efficiently."

# Accessing the modified docstring
print("Modified Docstring:\n")
print(calculate_miles.__doc__)
```

## Understanding Arbitrary Arguments (`*args` and `**kwargs`) in Python

Sometimes, you do not know in advance how many arguments will be passed into your function. Python provides two ways to handle an arbitrary number of arguments:

1. **Arbitrary Positional Arguments (`*args`)**: Allows a function to accept any number of positional arguments. Inside the function, they are received as a **tuple**.
2. **Arbitrary Keyword Arguments (`**kwargs`)**: Allows a function to accept any number of keyword arguments. Inside the function, they are received as a **dictionary**.

### 1. Arbitrary Positional Arguments (`*args`)
Use an asterisk (`*`) before the parameter name to collect an arbitrary number of positional arguments.

```python
def make_pizza(size, *toppings):
    """
    Summarize the pizza we are about to make.
    'size' is a regular positional argument.
    '*toppings' collects all remaining arguments into a tuple.
    """
    print(f"\nMaking a {size}-inch pizza with the following toppings:")
    for topping in toppings:
        print(f"- {topping}")

# Call the function with 3 toppings
make_pizza(12, "pepperoni", "mushrooms", "extra cheese")

# Call the function with 1 topping
make_pizza(16, "green peppers")
```

### 2. Arbitrary Keyword Arguments (`**kwargs`)
Use two asterisks (`**`) before the parameter name to collect an arbitrary number of keyword arguments.

```python
def build_profile(first, last, **user_info):
    """
    Build a dictionary containing everything we know about a user.
    'first' and 'last' are required positional arguments.
    '**user_info' collects all other keyword-value pairs into a dictionary.
    """
    user_info['first_name'] = first
    user_info['last_name'] = last
    return user_info

# Pass arbitrary keyword arguments to build the profile
user_profile = build_profile('Albert', 'Einstein',
                             location='Princeton',
                             field='Physics')
print(user_profile)
```
