**Example of an if, elif, and else branches:**

```python
def water_temperature(temperature):
  if (temperature == 0):
    print("Water is frozen.")
  elif (temperature == 100):
    print ("Water is boiling.")
  else:
    print ("Water temperature is " + str(temperature) + " degrees.")

water_temperature(0)
water_temperature(100)
water_temperature(50)
```
