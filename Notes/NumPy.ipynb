**NumPy tutorial**

This is a Python package that provides an alternative to Python lists, which is NumPy array.

#
**How to install?**

`pip3 install numpy`

#
**Example use follows:**

```python
import numpy as np

# We make two lists and then turn them into NumPy arrays.

left = [1,2,3,4,5]
left_array = np.array(left)
right = [6,7,8,9,10]
right_array = np.array(right)

# We can apply an operation like summing onto every element of the arrays.
sum_array = left_array + right_array
print(sum_array)

# Let's see the difference with adding two lists
print(left+right)

#subsetting
print(sum_array[1])
## Another way of subsetting
print(sum_array[sum_array<8])
```
