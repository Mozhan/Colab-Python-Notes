# Introduction to Pandas

**Pandas** is a fast, powerful, flexible, and easy-to-use open-source data analysis and manipulation tool built on top of the Python programming language.

In this tutorial, we will cover:
1. Importing pandas and creating DataFrames
2. Viewing and inspecting data
3. Selecting data (columns, rows, and filtering)
4. Basic data operations and aggregations

## 1. Importing Pandas and Creating Data

First, we import the library, conventionally aliased as `pd`. We will then create a simple DataFrame from a dictionary.

```python
import pandas as pd

# Creating a dictionary of sample data
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
    'Age': [24, 27, 22, 32, 29],
    'City': ['New York', 'Los Angeles', 'Chicago', 'Chicago', 'Austin'],
    'Salary': [70000, 85000, 54000, 120000, 95000]
}

# Converting dictionary into a pandas DataFrame
df = pd.DataFrame(data)

# Display the DataFrame
display(df)
```

## 2. Inspecting and Viewing Data

We can use quick helper methods to check the dimensions, statistics, and types of our dataset.

```python
# Display basic information about columns, non-null values, and data types
print("--- df.info() ---")
df.info()

# Display statistical summary of numerical columns
print("\n--- df.describe() ---")
display(df.describe())
```

## 3. Selecting and Filtering Data

Pandas makes it easy to isolate specific columns or rows based on conditional logic.

```python
# Selecting a single column (returns a Series)
print("Names column:")
display(df['Name'])

# Filtering rows where Age is greater than 25
print("\nEmployees older than 25:")
older_than_25 = df[df['Age'] > 25]
display(older_than_25)
```

## 4. Aggregation and Grouping

We can group our dataset by categorical variables and perform aggregate computations like average, sum, or count.

```python
# Grouping by City and finding the average salary in each city
avg_salary_by_city = df.groupby('City')['Salary'].mean().reset_index()

print("Average Salary by City:")
display(avg_salary_by_city)
```

## 5. Reading External Files with Pandas

One of the most common tasks in data analysis is loading external datasets. Pandas provides highly optimized functions for reading various file formats directly into DataFrames, such as:

* **CSV files (`pd.read_csv`)**
* **Excel files (`pd.read_excel`)**
* **JSON files (`pd.read_json`)**
* **SQL Databases (`pd.read_sql`)**

```python
# Example: How you would read a CSV file
# df_csv = pd.read_csv('path_to_file.csv')

# Example: How you would read an Excel file
# df_excel = pd.read_excel('path_to_file.xlsx', sheet_name='Sheet1')

# Let's create a temporary CSV file to demonstrate reading it
import tempfile

with tempfile.NamedTemporaryFile(mode='w+', delete=False, suffix='.csv') as temp_file:
    temp_file.write("id,item,price,in_stock\n1,Laptop,999.99,True\n2,Mouse,25.00,True\n3,Monitor,150.00,False\n")
    temp_file_path = temp_file.name

# Reading the CSV file back using pandas
loaded_df = pd.read_csv(temp_file_path)
display(loaded_df)
```

## 6. Storing Data with Pandas

Once you have cleaned or processed your data, you will often want to save it back to disk. Pandas provides straightforward methods to export your DataFrames:

* **CSV files (`df.to_csv`)**
* **Excel files (`df.to_excel`)**
* **JSON files (`df.to_json`)**

```python
# Let's export our filtered DataFrame (older_than_25) to a CSV file
# We set index=False so that pandas does not write the row index numbers into the file
older_than_25.to_csv('older_employees.csv', index=False)

# Let's verify by loading the newly created file back and displaying it
verify_df = pd.read_csv('older_employees.csv')
display(verify_df)
```
