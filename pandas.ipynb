# Introduction to Pandas

**Pandas** is a fast, powerful, flexible, and easy-to-use open-source data analysis and manipulation tool built on top of the Python programming language.

In this tutorial, we will cover:
1. Importing pandas and creating DataFrames
2. Viewing and inspecting data
3. Selecting data (columns, rows, and filtering)
4. Basic data operations and aggregations

## 1. Importing Pandas and Creating Data

First, we import the library, conventionally aliased as `pd`. We will then create a simple DataFrame from a dictionary.

```python
import pandas as pd

# Creating a dictionary of sample data
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
    'Age': [24, 27, 22, 32, 29],
    'City': ['New York', 'Los Angeles', 'Chicago', 'Chicago', 'Austin'],
    'Salary': [70000, 85000, 54000, 120000, 95000]
}

# Converting dictionary into a pandas DataFrame
df = pd.DataFrame(data)

# Display the DataFrame
display(df)
```

## 2. Inspecting and Viewing Data

We can use quick helper methods to check the dimensions, statistics, and types of our dataset.

```python
# Display basic information about columns, non-null values, and data types
print("--- df.info() ---")
df.info()

# Display statistical summary of numerical columns
print("\n--- df.describe() ---")
display(df.describe())
```

## 3. Selecting and Filtering Data

Pandas makes it easy to isolate specific columns or rows based on conditional logic.

```python
# Selecting a single column (returns a Series)
print("Names column:")
display(df['Name'])

# Filtering rows where Age is greater than 25
print("\nEmployees older than 25:")
older_than_25 = df[df['Age'] > 25]
display(older_than_25)
```

## 4. Aggregation and Grouping

We can group our dataset by categorical variables and perform aggregate computations like average, sum, or count.

```python
# Grouping by City and finding the average salary in each city
avg_salary_by_city = df.groupby('City')['Salary'].mean().reset_index()

print("Average Salary by City:")
display(avg_salary_by_city)
```
